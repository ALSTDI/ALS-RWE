# Day 2 Sample Notebook: Concept Sets and Data Quality

**OHDSI / OMOP Train-the-Trainer, ALS Therapy Development Institute**

Companion to the Day 2 module. You will build a concept set using descendants, then run one hand-written check from each Kahn data quality
category (conformance, completeness, plausibility) against a synthetic ALS CDM. These checks are all
verification checks: they compare the data with expectations that come from the data model itself. The
framework also describes validation against external benchmarks, which this notebook does not cover.

> **Site specific.** This notebook builds a synthetic CDM in memory so it runs in Colab with no credentials and no PHI. To use real data, replace the SQLite connection with your site's governed CDM connection (Databricks, Snowflake, Postgres, BigQuery, and so on). The SQL logic is the same; only the connection changes. Not everyone uses Databricks.

### Objectives
1. Build a concept set with descendants and validate it with SQL.
2. Run a conformance check (do concept IDs exist and belong to the right domain?).
3. Run completeness checks (how many records are unmapped, and how many persons lack a measurement?).
4. Run a plausibility check (are values believable?).
5. See why a score with a LOINC code, the ALSFRS-R, can still be missing from the structured data.

> **About the concept IDs.** 373182 is the OMOP concept ID for amyotrophic lateral sclerosis used on the ALS TDI OMOP data set page. The other drug, procedure, measurement, and observation concept IDs and codes in this synthetic vocabulary are placeholders for teaching and were not checked against Athena. Look up the real concept IDs in [Athena](https://athena.ohdsi.org/) before using any of these queries against a real CDM.


## Step 0: Build the synthetic CDM

````python
import sqlite3, numpy as np, pandas as pd
np.random.seed(7)

def build_als_cdm(n=250):
    """Build a tiny SYNTHETIC ALS-flavored OMOP CDM in memory. No real data, no PHI."""
    con = sqlite3.connect(":memory:"); cur = con.cursor()
    cur.executescript("""
    CREATE TABLE concept(concept_id INT, concept_name TEXT, domain_id TEXT, vocabulary_id TEXT,
        concept_class_id TEXT, standard_concept TEXT, concept_code TEXT);
    CREATE TABLE concept_ancestor(ancestor_concept_id INT, descendant_concept_id INT, min_levels_of_separation INT);
    CREATE TABLE person(person_id INT, gender_concept_id INT, year_of_birth INT);
    CREATE TABLE observation_period(person_id INT, observation_period_start_date TEXT, observation_period_end_date TEXT);
    CREATE TABLE condition_occurrence(person_id INT, condition_concept_id INT, condition_start_date TEXT);
    CREATE TABLE drug_exposure(person_id INT, drug_concept_id INT, drug_exposure_start_date TEXT);
    CREATE TABLE measurement(person_id INT, measurement_concept_id INT, measurement_date TEXT, value_as_number REAL);
    CREATE TABLE procedure_occurrence(person_id INT, procedure_concept_id INT, procedure_date TEXT);
    """)
    concepts=[
     (373182,"Amyotrophic lateral sclerosis","Condition","SNOMED","Clinical Finding","S","86044005"),
     (4134454,"Bulbar onset","Observation","SNOMED","Clinical Finding","S","230258005"),
     (1314865,"riluzole","Drug","RxNorm","Ingredient","S","9468"),
     (19077572,"riluzole 50 MG Oral Tablet","Drug","RxNorm","Clinical Drug","S","316158"),
     (1326727,"edaravone","Drug","RxNorm","Ingredient","S","1546020"),
     (40234834,"edaravone 30 MG/100mL","Drug","RxNorm","Clinical Drug","S","1546022"),
     (9990001,"sodium phenylbutyrate / taurursodiol","Drug","RxNorm","Ingredient","S","2630918"),
     (9990002,"tofersen","Drug","RxNorm","Ingredient","S","2629000"),
     (1304643,"baclofen","Drug","RxNorm","Ingredient","S","1292"),
     (4052536,"Gastrostomy","Procedure","SNOMED","Procedure","S","36245001"),
     (3024171,"Forced vital capacity","Measurement","LOINC","Clinical Obs","S","19868-9"),
     (8532,"FEMALE","Gender","Gender","Gender","S","F"),(8507,"MALE","Gender","Gender","Gender","S","M"),
     (0,"No matching concept","Metadata","None","Undefined",None,"0")]
    cur.executemany("INSERT INTO concept VALUES(?,?,?,?,?,?,?)",concepts)
    cur.executemany("INSERT INTO concept_ancestor VALUES(?,?,?)",[
     (1314865,19077572,1),(1314865,1314865,0),(1326727,40234834,1),(1326727,1326727,0),
     (9990001,9990001,0),(9990002,9990002,0),(1304643,1304643,0)])
    persons=[];obsp=[];cond=[];drug=[];meas=[];proc=[]
    for pid in range(1,n+1):
        sex=int(np.random.choice([8532,8507])); yob=int(np.random.normal(1958,11))
        persons.append((pid,sex,yob))
        dx_year=np.random.randint(2018,2024); dx=f"{dx_year}-{np.random.randint(1,13):02d}-15"
        cond.append((pid,373182,dx))
        bulbar=np.random.rand()<0.3
        if bulbar: cond.append((pid,4134454,dx))
        prior=int(np.random.choice([400,500,800,200,100],p=[.4,.2,.2,.1,.1]))
        start=pd.Timestamp(dx)-pd.Timedelta(days=prior)
        end=pd.Timestamp(dx)+pd.Timedelta(days=int(np.random.randint(365,1500)))
        obsp.append((pid,start.date().isoformat(),end.date().isoformat()))
        base_fvc=float(np.clip(np.random.normal(85-(10 if bulbar else 0),12),30,110))
        r=np.random.rand()
        if r<0.08: pass                                  # missing (completeness)
        elif r<0.11: meas.append((pid,3024171,dx,float(np.random.choice([0,250.0]))))  # implausible (plausibility)
        else: meas.append((pid,3024171,dx,round(base_fvc,1)))
        idx=pd.Timestamp(dx)+pd.Timedelta(days=int(np.random.randint(0,40)))
        if np.random.rand()<0.85:
            drug.append((pid,19077572,idx.date().isoformat()))
            if np.random.rand()<0.45:
                drug.append((pid,40234834,(idx+pd.Timedelta(days=int(np.random.randint(60,300)))).date().isoformat()))
        else:
            drug.append((pid,40234834,idx.date().isoformat()))
        if np.random.rand()<0.15:
            drug.append((pid,9990001,(idx+pd.Timedelta(days=int(np.random.randint(100,500)))).date().isoformat()))
        if bulbar and np.random.rand()<0.4: drug.append((pid,1304643,idx.date().isoformat()))
        if np.random.rand()<0.05: drug.append((pid,0,idx.date().isoformat()))   # unmapped (completeness)
        age=dx_year-yob
        logit=-2.0+0.04*(age-60)+1.1*bulbar-0.03*(base_fvc-80)
        if np.random.rand()<1/(1+np.exp(-logit)):
            proc.append((pid,4052536,(pd.Timestamp(dx)+pd.Timedelta(days=int(np.random.randint(60,360)))).date().isoformat()))
    cur.executemany("INSERT INTO person VALUES(?,?,?)",persons)
    cur.executemany("INSERT INTO observation_period VALUES(?,?,?)",obsp)
    cur.executemany("INSERT INTO condition_occurrence VALUES(?,?,?)",cond)
    cur.executemany("INSERT INTO drug_exposure VALUES(?,?,?)",drug)
    cur.executemany("INSERT INTO measurement VALUES(?,?,?,?)",meas)
    cur.executemany("INSERT INTO procedure_occurrence VALUES(?,?,?)",proc)
    con.commit(); return con

con = build_als_cdm()
def q(sql): return pd.read_sql_query(sql, con)
print("Synthetic ALS CDM ready:", q("SELECT COUNT(*) n FROM person").n[0], "persons")
````

## Step 1: A concept set with descendants
A concept set for the riluzole ingredient should capture every product below it. The `concept_ancestor`
table makes this one join instead of listing each product.

````python
q("""
SELECT ing.concept_name AS ingredient, d.concept_name AS product_in_set
FROM concept_ancestor ca
JOIN concept ing ON ing.concept_id = ca.ancestor_concept_id
JOIN concept d   ON d.concept_id   = ca.descendant_concept_id
WHERE ca.ancestor_concept_id = 1314865
""")
````

## Step 2: Conformance
Does every `drug_concept_id` exist in the `concept` table and belong to the Drug domain? DQD calls these checks
`isForeignKey` (relational conformance) and `fkDomain` (value conformance). Rows with `concept_id = 0` are set
aside here, because DQD counts them under completeness (Step 3).

````python
q("""
SELECT CASE WHEN c.concept_id IS NULL THEN 'not in concept table'
            WHEN c.domain_id <> 'Drug' THEN 'wrong domain'
            ELSE 'conforms' END AS status,
       COUNT(*) AS rows
FROM drug_exposure de
LEFT JOIN concept c ON c.concept_id = de.drug_concept_id
WHERE de.drug_concept_id <> 0
GROUP BY status
""")
````

## Step 3: Completeness
Are values present where expected? Two checks:

- The percent of drug records stored with `concept_id = 0`, which DQD calls `standardConceptRecordCompleteness`.
- The percent of persons with no FVC measurement. This is a study-specific check, similar in spirit to DQD's
  `measurePersonCompleteness`, which counts persons with no record in a table.

````python
print(q("""
SELECT SUM(CASE WHEN drug_concept_id = 0 THEN 1 ELSE 0 END) AS unmapped_rows,
       COUNT(*) AS total_rows,
       ROUND(100.0 * SUM(CASE WHEN drug_concept_id = 0 THEN 1 ELSE 0 END) / COUNT(*), 1) AS pct_unmapped
FROM drug_exposure
"""))

q("""
SELECT COUNT(*) AS als_persons,
       SUM(CASE WHEN m.person_id IS NULL THEN 1 ELSE 0 END) AS missing_fvc
FROM person p
LEFT JOIN measurement m
  ON m.person_id = p.person_id AND m.measurement_concept_id = 3024171
""")
````

## Step 4: Plausibility
The synthetic FVC values in this notebook are percent of predicted, so a value of 0 or a value above 120 is
treated as implausible. This is atemporal plausibility; DQD's `plausibleValueLow` and `plausibleValueHigh`
work the same way. The cut-offs are teaching values for this synthetic data set, not clinical thresholds.

````python
q("""
SELECT value_as_number,
       CASE WHEN value_as_number <= 0 OR value_as_number > 120 THEN 'implausible' ELSE 'ok' END AS flag
FROM measurement
WHERE measurement_concept_id = 3024171 AND (value_as_number <= 0 OR value_as_number > 120)
""")
````

## Step 5: A score that exists in the vocabulary and is often kept in notes
The ALSFRS-R has LOINC codes for the scale (82954-9), the total score (82953-1), and each item, so the concept exists
at every site. At many sites the scores are written in clinic notes and never reach a structured table. This step adds
a synthetic `observation` table and a synthetic `note` table to show what that looks like in a completeness check.

The share of people with a structured score and the share with a score in a note are made up for this exercise. The
concept ID used for the total score, 42529084, is the last ID in the range the ALS TDI data set documents for the
ALSFRS-R (42529071 to 42529084); confirm in Athena which ID is the total score.

````python
cur = con.cursor()
cur.executescript("""
CREATE TABLE observation(person_id INT, observation_concept_id INT, observation_date TEXT, value_as_number REAL);
CREATE TABLE note(person_id INT, note_date TEXT, note_text TEXT);
""")
cur.execute("INSERT INTO concept VALUES(?,?,?,?,?,?,?)",
            (42529084, "Total score [ALSFRS-R]", "Observation", "LOINC", "Survey", "S", "82953-1"))
rng = np.random.default_rng(11)
people = q("""SELECT p.person_id, c.condition_start_date AS dx
              FROM person p JOIN condition_occurrence c
                ON c.person_id = p.person_id AND c.condition_concept_id = 373182""")
for pid, dx in people.itertuples(index=False):
    score = int(np.clip(rng.normal(36, 6), 10, 48))
    if rng.random() < 0.30:        # recorded as structured data
        cur.execute("INSERT INTO observation VALUES(?,?,?,?)", (int(pid), 42529084, dx, score))
    elif rng.random() < 0.80:      # written in a clinic note instead
        cur.execute("INSERT INTO note VALUES(?,?,?)",
                    (int(pid), dx, f"Seen in ALS clinic. ALSFRS-R total {score}/48. Plan reviewed."))
con.commit()

total      = q("SELECT COUNT(*) n FROM person").n[0]
structured = q("SELECT COUNT(DISTINCT person_id) n FROM observation WHERE observation_concept_id = 42529084").n[0]
in_notes   = q("SELECT COUNT(DISTINCT person_id) n FROM note WHERE LOWER(note_text) LIKE '%alsfrs%'").n[0]
print(f"Persons:                               {total}")
print(f"With a structured ALSFRS-R total:      {structured} ({100*structured/total:.0f}%)")
print(f"With the score only in a clinic note:  {in_notes} ({100*in_notes/total:.0f}%)")
````

A query on the standard concept finds only the structured records. The rest of the scores are in the data, in a
place a concept query does not look. Text extraction can recover them; in an OMOP CDM the extracted values would be
written to `NOTE_NLP` or loaded as structured records. The line below pulls the number out of the synthetic notes.

````python
notes = q("SELECT person_id, note_text FROM note")
notes["score_from_note"] = notes.note_text.str.extract(r"ALSFRS-R total (\d+)/48").astype(float)
notes.head()
````
